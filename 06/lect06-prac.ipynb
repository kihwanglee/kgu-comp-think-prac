## 1. 리스트 만들기와 인덱싱

**리스트**는 여러 값을 하나의 이름으로 묶는 자료형입니다. 대괄호 `[]` 안에 값을 쉼표로 나열합니다. 각 값은 **0부터** 시작하는 인덱스로 꺼냅니다.

````python
scores = [90, 85, 70, 95]
print(scores)
print(scores[0])    # 90  (첫 번째)
print(scores[-1])   # 95  (마지막)
````

**슬라이싱** — `리스트[시작:끝]` 으로 일부를 잘라냅니다. 시작은 포함, 끝은 제외.

````python
scores = [90, 85, 70, 95]
print(scores[1:3])   # [85, 70]
print(scores[::-1])  # [95, 70, 85, 90]  (역순)
````

### 수정·길이·메서드

인덱스로 값을 바꾸고, `len()`으로 개수를, 메서드로 추가·삭제를 합니다.

````python
scores = [90, 85, 70, 95]
scores[2] = 100          # 세 번째 값 수정
print(scores, len(scores))

fruits = ["사과", "바나나"]
fruits.append("포도")     # 맨 뒤에 추가
fruits.remove("바나나")   # 값으로 삭제
print(fruits)            # ['사과', '포도']
print("사과" in fruits)  # True
````

## 2. `for` 반복문과 `range()`

`for 변수 in range(횟수):` 는 정해진 횟수만큼 반복합니다. `range(n)`은 `0`~`n-1`, `range(시작, 끝)`은 시작~끝 직전, `range(시작, 끝, 간격)`으로 건너뛰거나 거꾸로 셉니다.

````python
for i in range(5):
    print(i)        # 0 1 2 3 4
````

````python
for i in range(1, 6):
    print(i)        # 1 2 3 4 5

for i in range(5, 0, -1):
    print(i)        # 5 4 3 2 1 (역순)
````

## 3. 리스트와 `for` 문

`for 변수 in 리스트:` 로 리스트의 값을 하나씩 꺼내 반복합니다.

````python
fruits = ["사과", "바나나", "포도"]
for fruit in fruits:
    print(f"나는 {fruit}를 좋아합니다.")
````

**합계와 평균** — 합을 담을 변수를 `0`에서 시작해 하나씩 더합니다.

````python
scores = [90, 85, 70, 95]
total = 0
for score in scores:
    total += score
print(f"합계: {total}, 평균: {total / len(scores):.1f}")
print(sum(scores), max(scores), min(scores))
````

**조건에 맞는 값만** — `for` 안에 `if`를 써서 60점 이상만 셉니다.

````python
scores = [90, 55, 70, 40, 85]
count = 0
for score in scores:
    if score >= 60:
        count += 1
print(f"60점 이상: {count}명")
````

## 4. 패턴·중첩 반복·구구단

문자열 반복(`*`)과 `for`로 도형을, 중첩 `for`로 구구단을 만듭니다.

**별 찍기** — 반복 변수가 커질수록 별이 늘어납니다.

````python
for i in range(1, 6):
    print("*" * i)
````

**중첩 `for`** — 바깥 반복이 한 번 돌 때 안쪽 반복이 전부 돕니다.

````python
for i in range(1, 4):       # 바깥: 1, 2, 3
    for j in range(1, 4):   # 안쪽: 1, 2, 3
        print(i, j)
````

**구구단 2단** — 한 단을 먼저 출력해 봅니다.

````python
dan = 2
for i in range(1, 10):
    print(f"{dan} x {i} = {dan * i}")
````
