## 1. 산술 연산자 복습

나눗셈은 `/`(실수), `//`(몫), `%`(나머지)로 나뉩니다. `**`는 지수입니다.

````python
print(7 / 4)    # 1.75
print(7 // 4)   # 1
print(7 % 4)    # 3
print(2 ** 10)  # 1024
````

`%`(나머지)는 짝수·홀수 판별에 자주 쓰입니다. `수 % 2`가 `0`이면 짝수입니다.

````python
print(10 % 2)   # 0 → 짝수
print(7 % 2)    # 1 → 홀수
````

## 2. 관계 연산자

두 값을 비교하여 참(`True`)/거짓(`False`)을 돌려줍니다: `==`, `!=`, `>`, `<`, `>=`, `<=`.

````python
print(7 > 4)    # True
print(7 < 4)    # False
print(5 == 5)   # True
print(5 != 5)   # False
````

문자열도 `==`, `!=`로 비교할 수 있습니다. 대소문자는 구분됩니다.

````python
print("python" == "python")   # True
print("Cat" == "cat")         # False (대소문자 구분)
````

> ⚠️ `=`는 **값을 저장(할당)**, `==`는 **같은지 비교**. 서로 다릅니다!

## 3. 논리 연산자

`and`(둘 다 참), `or`(하나라도 참), `not`(뒤집기)으로 여러 조건을 조합합니다.

````python
print(True and False)   # False
print(True or False)    # True
print(not True)         # False
````

점수가 80 이상이고 90 미만인지 `and`로 확인합니다.

````python
score = 85
print(score >= 80 and score < 90)   # True
````

## 4. 조건문: `if` 와 `if`-`else`

`if 조건:` 다음 줄을 **들여쓰기(4칸)**하여 조건이 참일 때 실행할 문장을 씁니다. `else`는 거짓일 때 실행됩니다.

````python
age = 20
if age >= 19:
    print("성인입니다.")
print("프로그램 끝")
````

````python
number = 7
if number % 2 == 0:
    print("짝수입니다.")
else:
    print("홀수입니다.")
````

## 5. `if`-`elif`-`else`

조건이 셋 이상이면 `elif`로 갈래를 늘립니다. 위에서부터 검사하여 **처음 참인 곳만** 실행합니다.

````python
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
print(f"학점은 {grade}입니다.")
````

## 6. 종합 활용

배운 것을 엮어 실전 판단 프로그램을 만듭니다.

**윤년 판별** — 4의 배수이면서 (100의 배수가 아니거나 400의 배수). `2024`, `2100`을 입력해 보세요.

````python
year = input("연도: ")
year = int(year)
if year % 4 == 0 and (year % 100 != 0 or year % 400 == 0):
    print(f"{year}년은 윤년입니다.")
else:
    print(f"{year}년은 평년입니다.")
````

**세 수 중 최댓값** — 세 수를 입력받아 가장 큰 수를 찾습니다.

````python
a = input("첫째 수: ")
a = int(a)
b = input("둘째 수: ")
b = int(b)
c = input("셋째 수: ")
c = int(c)
if a >= b and a >= c:
    biggest = a
elif b >= c:
    biggest = b
else:
    biggest = c
print(f"가장 큰 수는 {biggest}입니다.")
````
