## 2. 간단한 계산 실습

파이썬은 강력한 계산기입니다. 주요 산술 연산자는 다음과 같습니다.

| 연산자 | 의미 | 예 | 결과 |
|:---:|:---:|:---:|:---:|
| `+` `-` `*` | 덧셈·뺄셈·곱셈 | `3 * 4` | `12` |
| `/` | 나눗셈(실수) | `7 / 2` | `3.5` |
| `//` | 몫(정수) | `7 // 2` | `3` |
| `%` | 나머지 | `7 % 2` | `1` |
| `**` | 거듭제곱 | `2 ** 8` | `256` |

````python
print(3 + 2)
print(3 * 4)
print(2 ** 8)
````

나눗셈은 세 종류가 있습니다: `/`(실수 나눗셈), `//`(정수 몫), `%`(나머지).

````python
print(7 / 2)   # 3.5
print(7 // 2)  # 3
print(7 % 2)   # 1
````

> ⚠️ **주의**: `/` 의 결과는 나누어떨어져도 항상 **실수**입니다. `10 / 5` 는 `2` 가 아니라 `2.0`.

곱셈·나눗셈이 덧셈·뺄셈보다 먼저 계산됩니다. 순서를 바꾸려면 괄호를 씁니다.

````python
print(3 + 2 * 4)    # 11
print((3 + 2) * 4)  # 20
````

## 3. 문제를 계산으로 표현하기

컴퓨팅 사고는 **일상의 문제를 컴퓨터가 처리할 수 있는 계산으로 표현**하는 데서 출발합니다. 문제를 잘게 나누어(분해) 계산식으로 옮겨 봅시다.

**화씨 온도를 섭씨로** — 공식: 섭씨 = (화씨 − 32) × 5 ÷ 9

````python
print((77 - 32) * 5 / 9)   # 화씨 77도 → 섭씨
````

**초를 분과 초로** — 몫(`//`)과 나머지(`%`) 활용

````python
print(1000 // 60)   # 분
print(1000 % 60)    # 초
````
