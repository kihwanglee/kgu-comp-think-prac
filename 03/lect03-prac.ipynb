## 개념: 변수는 '이름'과 '값'의 연결

컴퓨터 세계에서 어떤 대상을 다루려면 **이름(name)·값(value)·유형(type)**이 정해져야 합니다. 그런데 **유형은 값에 딸린 속성**(예: `90`은 정수, `"홍길동"`은 문자열)이라, 실제로 하는 일은 **이름에 값을 연결(할당)**하는 것입니다.

자연언어의 **단어**가 **형식(소리·글자)**과 **의미**가 짝을 이뤄야 유효하듯, 변수도 **이름(형식)**에 **값(의미)**이 연결돼야 유효합니다.

> **형식(소리·글자) ↔ 의미**  ≈  **이름(name) ↔ 값(value)**

## 1. 변수 만들기와 사용

`=` 는 오른쪽 값을 왼쪽 이름에 **저장(할당)**하라는 기호입니다.

````python
score = 90
print(score)
````

변수 여러 개를 만들어 계산에 쓸 수 있고, 자기 자신을 이용해 값을 바꿀 수도 있습니다.

````python
x = 100
y = 200
total = x + y
print(total)   # 300
````

````python
count = 5
count = count + 1
print(count)   # 6
````

## 2. 자료형(data type)

값에는 종류(자료형)가 있습니다. 파이썬의 기본 자료형 네 가지:

| 자료형 | 이름 | 예 |
|:---:|:---:|:---|
| 정수 | `int` | `90`, `-3` |
| 실수 | `float` | `3.5`, `1.25` |
| 문자열 | `str` | `"안녕"`, `'파이썬'` |
| 불리언 | `bool` | `True`, `False` |

`type()` 으로 값의 자료형을 확인할 수 있습니다.

````python
print(type(90))
print(type(3.5))
print(type("안녕"))
print(type(True))
````

**불리언(bool)** 은 참(`True`)/거짓(`False`) 둘 중 하나입니다. 두 값을 비교하면 결과가 불리언으로 나옵니다.

````python
print(3 > 2)
print(10 == 5)
````

자료형은 필요하면 `str()`(→문자열), `int()`(→정수)로 바꿉니다.

````python
print("나이는 " + str(20) + "살")
print(int("15") + int("5"))
````

## 3. 사용자로부터 입력 받기: `input()`

`input("안내 문구")` 는 사용자가 입력한 값을 **문자열**로 돌려줍니다.

> 실행하면 입력창이 나타납니다. 값을 입력하고 `Enter` 를 누르세요.

````python
name = input("이름을 입력하시오: ")
print(name, "님, 반갑습니다!")
````

`input()` 은 **항상 문자열**을 주므로, 계산하려면 `int()` 로 정수로 바꿉니다. 입력 문장과 변환 문장을 두 줄로 나눠 씁니다.

````python
x = input("첫 번째 정수를 입력하시오: ")
x = int(x)
y = input("두 번째 정수를 입력하시오: ")
y = int(y)
print(x, "과", y, "의 합은", x + y, "입니다.")
````

**나이 계산기** — 태어난 해를 입력받아 올해(2026년) 기준 나이를 구합니다.

````python
year = input("태어난 해를 입력하시오: ")
year = int(year)
age = 2026 - year
print(f"당신의 나이는 {age}살입니다.")
````

## 4. f-문자열로 깔끔하게 출력하기

**f-문자열**은 문자열 앞에 `f` 를 붙이고 `{}` 안에 변수를 넣어, 문장 속에 값을 끼워 넣는 방법입니다.

````python
name = "홍길동"
age = 20
print(f"{name}님은 {age}살입니다.")
````

`{값:.2f}` 처럼 쓰면 **소수점 아래 자릿수**를 지정할 수 있습니다. (`.2f` 는 둘째 자리까지)

````python
pi = 3.141592
print(f"원주율은 약 {pi:.2f} 입니다.")
````

## 5. 종합 예제: 간단한 입출력 프로그램

문제를 **입력 → 처리 → 출력** 으로 나누어 생각하면 프로그램이 자연스럽게 만들어집니다.

**평균 계산기** — 두 과목 점수를 입력받아 평균을 소수점 첫째 자리까지 출력합니다.

````python
korean = input("국어 점수: ")
korean = int(korean)
english = input("영어 점수: ")
english = int(english)
average = (korean + english) / 2
print(f"평균은 {average:.1f}점입니다.")
````

**BMI 계산기** — 키(m)와 몸무게(kg)를 입력받아 BMI를 구합니다. (소수가 있으므로 `float()`)

````python
height = input("키(m)를 입력하시오: ")
height = float(height)
weight = input("몸무게(kg)를 입력하시오: ")
weight = float(weight)
bmi = weight / (height * height)
print(f"당신의 BMI는 {bmi:.1f} 입니다.")
````
