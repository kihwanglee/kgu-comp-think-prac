## 1. 관계 연산자

두 값을 비교하여 참(`True`)/거짓(`False`)을 돌려줍니다: `==`, `!=`, `>`, `<`, `>=`, `<=`. 문자열도 비교할 수 있습니다.

````python
print(7 > 4)          # True
print(5 == 5)         # True
print(5 != 5)         # False
print("Cat" == "cat") # False (대소문자 구분)
````

> ⚠️ `=`는 **값을 저장(할당)**, `==`는 **같은지 비교**. 서로 다릅니다!

**산수 퀴즈** — 답이 맞는지 관계 연산자로 확인합니다.

````python
answer = input("7 × 8 = ? ")
answer = int(answer)
print(answer == 56)   # 맞으면 True
````

## 2. 논리 연산자

`and`(둘 다 참), `or`(하나라도 참), `not`(뒤집기)으로 여러 조건을 조합합니다.

````python
print(True and False)   # False
print(True or False)    # True
print(not (3 > 5))      # True
````

**놀이공원 탑승** — 키가 120cm 이상이고 200cm 이하여야 탈 수 있습니다. `and`로 묶습니다.

````python
height = input("키(cm): ")
height = int(height)
if height >= 120 and height <= 200:
    print("탑승할 수 있습니다.")
else:
    print("탑승할 수 없습니다.")
````

## 3. 조건문: `if` 와 `if`-`else`

`if 조건:` 다음 줄을 **들여쓰기(4칸)**하여, 조건이 참일 때 실행할 문장을 씁니다. `else`는 거짓일 때 실행됩니다.

````python
number = 7
if number % 2 == 0:
    print("짝수입니다.")
else:
    print("홀수입니다.")
````

**동전 던지기 게임** — `random` 모듈로 0/1을 무작위로 뽑습니다. `random`은 기본 내장 모듈이라 `import`만 하면 됩니다. 실행할 때마다 결과가 달라집니다.

````python
import random
coin = random.randint(0, 1)   # 0 또는 1
if coin == 0:
    print("앞면입니다.")
else:
    print("뒷면입니다.")
````

## 4. `if`-`elif`-`else`

조건이 셋 이상이면 `elif`로 갈래를 늘립니다. 위에서부터 검사하여 **처음 참인 곳만** 실행합니다.

````python
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
print(f"학점은 {grade}입니다.")
````

**교통 신호등** — 문자열도 `==`로 비교해 갈래를 나눕니다.

````python
signal = input("신호등 색(빨강/노랑/초록): ")
if signal == "빨강":
    print("멈추세요.")
elif signal == "노랑":
    print("주의하세요.")
elif signal == "초록":
    print("건너세요.")
else:
    print("알 수 없는 신호입니다.")
````

**사칙연산 계산기** — 두 수와 연산 기호를 입력받아 기호에 따라 다른 계산을 합니다.

````python
a = input("첫째 수: ")
a = int(a)
op = input("연산 기호(+, -, *, /): ")
b = input("둘째 수: ")
b = int(b)
if op == "+":
    print(a + b)
elif op == "-":
    print(a - b)
elif op == "*":
    print(a * b)
elif op == "/":
    print(a / b)
else:
    print("알 수 없는 연산입니다.")
````

## 5. 중첩 조건문과 판단 알고리즘 설계

조건문 안에 조건문을 넣어 **단계적으로** 판단합니다. 판단이 여러 갈래인 문제는 **플로우차트로 먼저 설계**한 뒤 코드로 옮깁니다.

**로그인 (중첩)** — 아이디가 맞은 다음에 비밀번호를 확인합니다.

````python
user_id = input("아이디: ")
if user_id == "kgu":
    password = input("비밀번호: ")
    if password == "1234":
        print("로그인 성공")
    else:
        print("비밀번호가 틀렸습니다.")
else:
    print("없는 아이디입니다.")
````

**세 수 중 가장 큰 수** — 여러 값을 비교해 하나를 고릅니다.

````python
a = input("첫째 수: ")
a = int(a)
b = input("둘째 수: ")
b = int(b)
c = input("셋째 수: ")
c = int(c)
if a >= b and a >= c:
    biggest = a
elif b >= c:
    biggest = b
else:
    biggest = c
print(f"가장 큰 수는 {biggest}입니다.")
````

## 6. 반복문 맛보기: `for` 문

같은 일을 여러 번 해야 할 때 문장을 복사하는 대신 **반복문**을 씁니다. `for 변수 in range(횟수):` 는 정해진 횟수만큼 들여쓴 문장을 반복합니다. (자세한 내용은 6강에서!)

````python
for i in range(3):
    print("안녕")
````

`range(3)`은 `0, 1, 2`를 차례로 만들어 냅니다. 반복 변수 `i`에 그 값이 하나씩 들어갑니다.

````python
for i in range(5):
    print(i)     # 0 1 2 3 4 를 한 줄에 하나씩
````

**반복과 조건을 함께** — `for`로 값을 하나씩 꺼내며 `if`로 골라냅니다. 1~10 중 짝수만 출력합니다.

````python
for i in range(1, 11):
    if i % 2 == 0:
        print(i)   # 2 4 6 8 10
````
