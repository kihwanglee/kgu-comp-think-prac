## 1. 코랩 노트북은 '셀'로 이루어져 있어요

컴퓨팅 사고는 **문제를 컴퓨터가 실행할 수 있는 절차로 표현**하는 것에서 시작합니다. 그 절차를 적고 실행하는 공간이 바로 코랩 노트북입니다.

노트북은 위에서 아래로 이어지는 **셀(cell)** 들의 묶음입니다.

| 셀 종류 | 하는 일 | 지금 이 셀은? |
|:---:|:---|:---|
| **텍스트 셀** | 제목·설명을 글로 씁니다 | 텍스트 셀입니다 |
| **코드 셀** | 파이썬 코드를 입력하고 **실행**합니다 | 아래에 있어요 ↓ |

- 셀을 한 번 클릭하면 **선택**, 두 번 클릭하면 **편집**할 수 있습니다.

### 셀을 실행하는 방법

아래 코드 셀을 클릭한 뒤, 다음 중 하나로 실행해 보세요.

- 셀 **왼쪽의 실행 버튼(▶)** 을 클릭
- 키보드로 **`Shift` + `Enter`** (실행하고 다음 셀로 이동)

> 💡 처음 실행하면 코랩이 클라우드에 파이썬 환경(런타임)을 준비하느라 **몇 초** 걸릴 수 있어요. 기다리면 됩니다.

````python
# 이 셀을 클릭하고 Shift+Enter 를 눌러 실행해 보세요.
print("Hello, World!")
````

방금 무슨 일이 일어났나요?

- `print(...)` 은 괄호 안의 내용을 **화면에 출력**하라는 명령(지시)입니다.
- 셀 **아래쪽에 결과** `Hello, World!` 가 나타났다면 성공입니다. 🎉
- 컴퓨터는 우리가 준 **명령을 순서대로 그대로 실행**합니다. 이 점이 컴퓨팅의 출발점이에요.

## 2. `print()` 으로 화면에 출력하기

`print()` 는 이 수업에서 가장 많이 쓰게 될 명령입니다. 괄호 안에 출력할 내용을 넣습니다.

- 글자(문자열)를 출력할 때는 반드시 **따옴표**로 감쌉니다. `"..."` 또는 `'...'` 둘 다 됩니다.

````python
print("안녕하세요!")
print('반갑습니다.')
````

### 값을 여러 개 함께 출력하기

`print()` 안에서 값을 **쉼표(`,`)** 로 구분하면, 값들 사이에 공백을 두고 함께 출력합니다.

````python
print("결과값은", 2 * 7, "입니다.")
````

- 따옴표로 감싼 `"결과값은"` 은 글자 그대로 출력되고,
- 따옴표가 없는 `2 * 7` 은 **계산되어** `14` 로 출력됩니다.
- 즉 컴퓨터는 **글자**와 **계산할 값**을 구분합니다.

## 3. 컴퓨터는 아주 빠른 계산기

컴퓨팅(computing)의 뿌리는 **계산(compute)** 입니다. 코드 셀에 수식을 넣고 실행하면 컴퓨터가 계산해 줍니다.

````python
print(3 + 2)
print(2 ** 10)
````

자주 쓰는 산술 연산자입니다.

| 연산자 | 의미 | 예시 | 결과 |
|:---:|:---:|:---:|:---:|
| `+` `-` `*` | 덧셈·뺄셈·곱셈 | `3 * 4` | `12` |
| `/` | 나눗셈(실수) | `7 / 2` | `3.5` |
| `**` | 거듭제곱 | `2 ** 10` | `1024` |

아래 셀에서 나눗셈과 거듭제곱도 확인해 보세요.

````python
print(7 / 2)
print(5 ** 3)
````

## 4. 컴퓨팅 사고 맛보기: 패턴 찾기

컴퓨터는 **규칙(패턴)** 을 아주 잘 다룹니다. 아래 계산을 실행하고, **어떤 규칙**이 있는지 살펴보세요.

````python
print(2 ** 1)
print(2 ** 2)
print(2 ** 3)
````

- 결과는 `2`, `4`, `8` 입니다. **앞의 결과에 2를 곱한 값**이 다음 결과가 되는 규칙이 보이나요?
- 이렇게 **반복되는 규칙을 찾아내는 것**을 컴퓨팅 사고에서 **패턴 인식**이라고 합니다.

### 🎯 도전 문제 3 (패턴 인식): 다음 값 예측하기

위 규칙대로라면 `2 ** 4` 의 결과는 얼마일까요? **먼저 머릿속으로 예측**한 뒤, 아래 셀에서 실행해 확인해 보세요.

````python
# 2 ** 4 를 출력해서 예측이 맞았는지 확인해 보세요.
````

<details><summary>💡 해답 보기</summary>

```python
print(2 ** 4)
```

결과는 `16` 입니다. `8 × 2 = 16`, 규칙 그대로죠? 패턴을 찾으면 **다음 값을 계산 없이도 예측**할 수 있습니다 — 이것이 컴퓨팅 사고의 힘입니다.

</details>
